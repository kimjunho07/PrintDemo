# Print 예제


```python
# 변수 선언
name = "Alice"
age = 25
score = 95.5
```

```python
# 1. 기본 출력
print("Hello, Python!")
```

```python
# 2. 여러 값 출력 (콤마로 구분 -> 자동 띄어쓰기)
print("Name:", name, "Age:", age, "Score:", score)
```

```python
# 3. f-string (가장 많이 쓰임, Python 3.6+)
print(f"My name is {name}, I am {age} years old, score: {score}")
```
